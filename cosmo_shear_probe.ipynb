```python
import numpy as np
import Cosmo_shear as cs
import matplotlib.pyplot as plt
```

```python
# Parametros fiduciales

Omega_b0_fid = 0.05
Omega_m0_fid = 0.32
h_fid = 0.67
ns_fid = 0.96
sigma8_fid = 0.816
Omega_DE0_fid = 0.68
w0_fid = -1.0
wa_fid = 0.0
gamma_fid = 0.55

#c = 9.72 * 10 ** (-15) # en Mpc # 300000 en km/s
c = 300000 #en km/s
Aia = 1.72
Cia = 0.0134
nia = -0.41
bia = 2.17
```

```python
L_array = np.log10(np.logspace(np.log10(10), np.log10(1500), 100)) # va de 0.00001 a 30 en espacios logaritmicos
zs = np.linspace(0.001, 2.5, 20) #np.logspace(np.log10(0.001), np.log10(2.5), 20)
cosmic_parametros = {'l': L_array , 'z': zs, 'type': 'standard', 'model' : 'ACDM_flat', 'IA': True}
cosmic = cs.CosmicShear(cosmic_parametros)
```

```python
k = cosmic.l_to_k(np.log10(100), 1, Omega_m0_fid, h_fid, Omega_b0_fid, Omega_DE0_fid, w0_fid, wa_fid, ns_fid, sigma8_fid, gamma_fid)
print(k)
```

```python
parametros = ['Omega_m0', 'h', 'Omega_b0', 'ns', 'sigma8']
```

```python
for parametro in parametros:
    der_P_par = np.array([float(cosmic.der_PPS_parametro(0.001, l, Omega_m0_fid, h_fid, Omega_b0_fid, Omega_DE0_fid, w0_fid, wa_fid, ns_fid, sigma8_fid, gamma_fid, parametro)) for l in L_array])
    plt.plot(L_array, der_P_par,)
    plt.xscale('log')
    plt.ylabel(r'$\partial P(k,z)/\partial$' + f' {parametro}' + ' [(Mpc/h)$^3$]')
    plt.xlabel('l')
    plt.show()
```

```python
C = cosmic.Cosmic_Shear(0, 0, Omega_m0_fid, h_fid, Omega_b0_fid, Omega_DE0_fid, w0_fid, wa_fid, ns_fid, sigma8_fid, gamma_fid, Aia, nia, bia)
plt.plot(10**L_array, C)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('l')
plt.ylabel('C(l) [(Mpc/h)$^3$]')        
```
